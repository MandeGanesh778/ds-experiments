```python
import pandas as pd
index=[
    ['Engineering','Engineering','mathematics','science'],
    ['AIML','DS','stastics','chemistry']
    ]
multi_index=pd.MultiIndex.from_arrays(index,names=['Department','branch'])
marks=pd.Series([85,78,92,88],index=multi_index)
print("Original series:")
print(marks)
print("\nData for Engineering:")
print(marks.loc['Engineering'])
print("\nData for AIML:")
print(marks.loc[('Engineering','AIML')])
print("\nData for science:")
print(marks.loc['science'])

```

```python
import pandas as pd
index=pd.MultiIndex.from_tuples(
    [
        ('Engineering','AIML'),
        ('Engineering','DS'),
        ('mathematics','stastics'),
        ('science','chemistry')
    ],
    names=['department','branch']
)
data=pd.DataFrame(
    {
        '2024':[78,85,98,88],
        '2025':[90,82,95,70]
    },
    index=index
)
print("Original tabular data")
print(data)
unstack_data=data.unstack()
print("\ndata after unstack():")
print(unstack_data)
stack_data=unstack_data.stack()
print("\n data after stack():")
print(stack_data)
```

```python
import pandas as pd
df1=pd.DataFrame(
    {
       'Name':['gani','bala','kushu'],
       'marks':[85,None,78],
       'Grade':['a','c','b']
    },
    index=[101,102,103]
)
df2=pd.DataFrame(
    {
        'Name':['shasi','vishnu','kani'],
        'marks':[99,100,98],
        'garde':['a','a','a']

    },
    index=[101,102,103]
)
print("First dataframe:")
print(df1)
print("Second dataframe:")
print(df2)
merged=pd.merge(
    df1,df2,
    left_index=True,
    right_index=True,
    how='outer',
    suffixes=('_df1','_df2')
)
print("\nMerged dataframe:")
print(merged)
combined=df1.combine_first(df2)
print("\n Combined dataframe:")
print(combined)

```
